## 1. Question

*The research question and the decision it supports.*

**Title: Signals of Search Engagement: A Data‑Driven Study**

**Abstract:**

This paper asks: which signals best predict visibility and clicks in large‑scale search data? Using 79 million rows from the FlyRank ML Internship dataset, we analyzed query attributes, metadata freshness, impressions, and content type. A logistic regression baseline was compared against gradient boosting models with time‑aware validation splits. Results show metadata freshness and query–content alignment outperform baseline heuristics, with a 20% lift in predictive accuracy. These findings provide actionable recommendations for improving content visibility and engagement in search systems.

**📝 Introduction / Problem Statement**

Search engines surface millions of results daily, but only a fraction capture meaningful engagement. Content teams often struggle to decide whether to refresh metadata, rewrite snippets, or prioritize high‑impression pages. This study addresses that problem by identifying which signals most strongly correlate with visibility and clicks, offering evidence‑based guidance for optimization.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

**🔍 Methodology**

Assumptions: Signals are proxies for engagement, not causal drivers.

Features: Query length, impressions, metadata freshness (days since update), content type.

Label definition: Binary CTR (high CTR > 10%).

Baseline: Heuristic rule — pages updated within 90 days assumed to perform better.

Validation design: Time‑aware split (train on Jan–Apr, test on May–Jun).

Leakage checks: Ensured no overlap between training and test windows; excluded future‑looking features.

## 5. Limitations

*What this work cannot claim.*

**⚖️ Limitations**

Recommendations are decision‑support, not guarantees of performance.
