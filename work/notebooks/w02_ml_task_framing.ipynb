## 3. Success metric

The success metric is **Normalized Discounted Cumulative Gain (NDCG)**.  
NDCG measures how well the ranking matches user preferences, rewarding correct ordering at the top of the list.  
Alternative metrics include **Mean Reciprocal Rank (MRR)** or **Precision@k**, but NDCG is the most standard for search.

## 5. Why ML beats a fixed rule here

A fixed rule (e.g., keyword overlap) is brittle:  
- It misses synonyms and semantic matches.  
- It cannot adapt to user behavior patterns.  
- It treats all queries the same, ignoring context.  

ML learns from **patterns in real user data**, improving relevance dynamically and outperforming static heuristics.
