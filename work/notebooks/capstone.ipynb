## 1. Question

*The research question and the decision it supports.*

Research Question: Which content signals are most strongly associated with higher visibility and clicks in FlyRank search data?  
Decision Supported: This work helps content teams decide where to focus — refreshing metadata, optimizing snippets, or monitoring high‑impression pages — to improve search visibility and engagement.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

Release: FlyRank ML Internship dataset (public Hugging Face release).

Tables used: impressions, clicks, metadata freshness, query attributes, content type.

Date window: January 2025 – June 2025.

Exclusions: private queries, client domains, raw exports, and credentials — excluded to keep the analysis public‑safe.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Assumptions: Signals are proxies for visibility/engagement, not causal drivers.

Features: query length, impressions, metadata freshness (days since update), content type.

Label definition: binary CTR (high CTR > 10%).

Baseline: heuristic — pages updated within 90 days assumed to perform better.

Validation design: time‑aware split (train on past, test on future).

Leakage checks: ensured no overlap between training and test windows; excluded future‑looking features.

## 5. Limitations

*What this work cannot claim.*

Correlation ≠ causation — signals are directional evidence only.

Dataset excludes private queries and client domains.

Recommendations are decision‑support, not guarantees of performance.

CTR threshold definition (>10%) is arbitrary and may not generalize.
